# Home Activity

## 1. Create five real-world problems from different domains

Five real-world problems from Education, Healthcare, Finance, Customer Support, and Programming were created and stored in the project dataset files.

**Location:** `dataset/home_activity_prompts.csv`

## 2. Solve each problem using Zero-shot prompting

The five problems were solved using Zero-shot prompting without providing examples.

**Location:** `dataset/home_activity_responses.csv`

## 3. Design One-shot prompts for all five problems and compare the responses

One-shot prompts were created for all five problems by providing one example for each problem. The generated responses were compared with the other prompting techniques.

**Location:** `dataset/home_activity_prompts.csv`  
**Responses:** `dataset/home_activity_responses.csv`

## 4. Generate responses using Few-shot prompting and evaluate improvements

Few-shot prompting was applied using multiple examples. The generated responses were evaluated based on clarity, relevance, and accuracy.

**Location:** `dataset/home_activity_responses.csv`  
**Evaluation:** `dataset/evaluation_results.csv`

## 5. Experiment with different roles using Role Prompting

Role Prompting was applied using the roles Teacher, Doctor, HR Manager, AI Engineer, and Travel Guide. The responses were generated according to the assigned roles.

**Location:** `dataset/home_activity_prompts.csv`  
**Responses:** `dataset/home_activity_responses.csv`

Three reasoning-based questions were created and tested using Chain-of-Thought prompting. The responses were compared with normal prompting.
